a) I created a function to calculate the percentage of late orders for a given promised delivery time. The function handles four cases: a specific zone and time block, all zones for a specific time block, a specific zone across all time blocks, and all zones across all time blocks.

```python
def late_percentage(zone, time_block, promise):

    if zone == "all" and time_block == "all":
        all_times = []

        for z in ZONES:
            for t in TIME_BLOCKS:
                times = delivery_times(z, t, promise, seed=1)
                all_times.extend(times)

        all_times = np.array(all_times)
        late_rate = np.mean(all_times > promise) * 100

    elif zone == "all":
        all_times = []

        for z in ZONES:
            times = delivery_times(z, time_block, promise, seed=1)
            all_times.extend(times)

        all_times = np.array(all_times)
        late_rate = np.mean(all_times > promise) * 100

    elif time_block == "all":
        all_times = []

        for t in TIME_BLOCKS:
            times = delivery_times(zone, t, promise, seed=1)
            all_times.extend(times)

        all_times = np.array(all_times)
        late_rate = np.mean(all_times > promise) * 100

    else:
        times = delivery_times(zone, time_block, promise, seed=1)
        late_rate = np.mean(times > promise) * 100

    return late_rate
```

b) I calculated the late arrival rate for each zone and time block using the current 45-minute promise and ranked the results in descending order. The results show which zone and time-block combinations have the highest late arrival rates.

```python
late_rates = []

for zone in ZONES:
    for time_block in TIME_BLOCKS:
        rate = late_percentage(zone, time_block, 45)
        late_rates.append((zone, time_block, rate))

late_rates.sort(key=lambda x: x[2], reverse=True)

for zone, time_block, rate in late_rates:
    print(zone, "-", time_block, ":", round(rate, 2), "%")
```

c) I created a function to calculate the average delivery time for a given promised delivery time. The function handles four cases: a specific zone and time block, all zones for a specific time block, a specific zone across all time blocks, and all zones across all time blocks.

```python
def average_delivery_time(zone, time_block, promise):

    if zone == "all" and time_block == "all":
        all_times = []

        for z in ZONES:
            for t in TIME_BLOCKS:
                times = delivery_times(z, t, promise, seed=1)
                all_times.extend(times)

        all_times = np.array(all_times)
        average_time = np.mean(all_times)

    elif zone == "all":
        all_times = []

        for z in ZONES:
            times = delivery_times(z, time_block, promise, seed=1)
            all_times.extend(times)

        all_times = np.array(all_times)
        average_time = np.mean(all_times)

    elif time_block == "all":
        all_times = []

        for t in TIME_BLOCKS:
            times = delivery_times(zone, t, promise, seed=1)
            all_times.extend(times)

        all_times = np.array(all_times)
        average_time = np.mean(all_times)

    else:
        times = delivery_times(zone, time_block, promise, seed=1)
        average_time = np.mean(times)

    return average_time
```

a) Each late order creates both a direct refund cost and an indirect cost from customer churn. I calculated the total cost per late order as the refund plus the profit lost from future churned orders.

```python
def late_order_cost(costs):
    refund = costs["refund"]
    churn_orders = costs["churn_orders"]
    margin = costs["margin"]

    total_cost = refund + churn_orders * margin

    return total_cost
```

test 2a

```python
cost_per_late_order = late_order_cost(COSTS)

print("Total cost per late order: $", round(cost_per_late_order, 2))
```

b) I first tested promises from 5 to 50 minutes in 5-minute increments. Since net profit was still increasing at the upper boundary of 50 minutes, I extended the search to 100 minutes. The net profit peaked at 55 minutes and then decreased as longer promises reduced the number of orders. Therefore, a wider range is useful because it ensures that the search passes the profit-maximizing point rather than stopping while profit is still increasing.

```python
times = delivery_times("Far West", "Fri/Sat eve", 45, seed=1)

number_of_orders = len(times)
number_of_late_orders = np.sum(times > 45)

order_profit = number_of_orders * COSTS["margin"]
late_cost = number_of_late_orders * late_order_cost(COSTS)

net_profit = order_profit - late_cost

print("Number of orders:", number_of_orders)
print("Number of late orders:", number_of_late_orders)
print("Profit from orders: $", round(order_profit, 2))
print("Late cost: $", round(late_cost, 2))
print("Net profit: $", round(net_profit, 2))
```

choose range: I first tested promised delivery times from 5 to 50 minutes in 5-minute increments. Since net profit was still increasing at the upper boundary of 50 minutes, this range might not contain the optimal promise. I therefore extended the search from 50 to 100 minutes. The net profit reached a peak at 55 minutes and then decreased as the promise became longer, suggesting that the expanded range successfully passed the profit-maximizing point.

```python
promises = list(range(5, 51, 5))

for promise in promises:
    times = delivery_times(
        "Far West", "Fri/Sat eve", promise, seed=1
    )

    number_of_orders = len(times)
    number_of_late_orders = np.sum(times > promise)

    order_profit = number_of_orders * COSTS["margin"]
    late_cost = number_of_late_orders * late_order_cost(COSTS)
    net_profit = order_profit - late_cost

    print(
        "Promise:", promise,
        "| Orders:", number_of_orders,
        "| Late orders:", number_of_late_orders,
        "| Net profit: $", round(net_profit, 2)
    )
```

find best promise

```python
def best_promise(zone, time_block, promises, costs):

    best_time = None
    best_profit = float("-inf")

    for promise in promises:
        times = delivery_times(
            zone, time_block, promise, seed=1
        )

        number_of_orders = len(times)
        number_of_late_orders = np.sum(times > promise)

        order_profit = number_of_orders * costs["margin"]
        late_cost = number_of_late_orders * late_order_cost(costs)

        net_profit = order_profit - late_cost

        if net_profit > best_profit:
            best_profit = net_profit
            best_time = promise

    return best_time, best_profit
```

```python
promises = list(range(5, 101, 5))

best_time, best_profit = best_promise(
    "Far West",
    "Fri/Sat eve",
    promises,
    COSTS
)

print("Best promise:", best_time, "minutes")
print("Best net profit: $", round(best_profit, 2))
```
